```python
prices = [100, 200, 300]
quantities = [2, 1, 3]
discount_rate = 10
tax_rate = 5
subtotal = sum(p * q for p, q in zip(prices, quantities))
discount = subtotal * discount_rate / 100
after_discount = subtotal - discount
tax = after_discount * tax_rate / 100
total_cost = after_discount + tax
print("Total cost:", total_cost)
```

```python
import pandas as pd
order_data = pd.DataFrame({
    "customer_id": [101, 102, 101, 103, 102],
    "order_date": ["2026-01-05", "2026-01-07", "2026-01-10", "2026-01-12", "2026-01-15"],
    "product_name": ["Laptop", "Phone", "Laptop", "Tablet", "Phone"],
    "order_quantity": [1, 2, 2, 1, 3]
})
total_orders = order_data.groupby("customer_id").size()
print("Total orders by each customer:")
print(total_orders)
average_quantity = order_data.groupby("product_name")["order_quantity"].mean()
print("\nAverage order quantity for each product:")
print(average_quantity)
order_data["order_date"] = pd.to_datetime(order_data["order_date"])
print("\nEarliest date:", order_data["order_date"].min())
print("Latest date:", order_data["order_date"].max())
```

```python
import pandas as pd
order_data = pd.DataFrame({
    "product_name": ["Laptop", "Phone", "Laptop", "Tablet", "Phone", "Laptop"],
    "order_quantity": [5, 10, 3, 8, 6, 7]
})
product_sales = order_data.groupby("product_name")["order_quantity"].sum()
top_5 = product_sales.sort_values(ascending=False).head(5)
print("Top 5 products sold:")
print(top_5)
```

```python
import pandas as pd
property_data = pd.DataFrame({
    "property_id": [101, 102, 103, 104, 105],
    "location": ["Chennai", "Bangalore", "Chennai", "Hyderabad", "Bangalore"],
    "bedrooms": [3, 5, 6, 4, 7],
    "area_sqft": [1200, 2500, 3000, 1800, 3500],
    "listing_price": [5000000, 8000000, 9000000, 6000000, 10000000]
})
average_price = property_data.groupby("location")["listing_price"].mean()
print("Average listing price:")
print(average_price)
count = (property_data["bedrooms"] > 4).sum()
print("\nProperties with more than 4 bedrooms:", count)
largest_property = property_data.loc[property_data["area_sqft"].idxmax()]
print("\nProperty with largest area:")
print(largest_property)
```
