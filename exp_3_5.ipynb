```python
import numpy as np
house_data = np.array([
    [3, 1500, 250000],
    [5, 2200, 400000],
    [6, 2500, 500000],
    [4, 1800, 300000],
    [7, 3000, 600000]
])
selected_houses = house_data[house_data[:, 0] > 4]
average_price = np.mean(selected_houses[:, 2])
print("Average sale price:", average_price)
```

```python
import numpy as np
sales_data = np.array([100000, 120000, 150000, 180000])
total_sales = np.sum(sales_data)
percentage_increase = ((sales_data[3] - sales_data[0]) / sales_data[0]) * 100
print("Total sales:", total_sales)
print("Percentage increase:", percentage_increase, "%")
```

```python
import numpy as np
fuel_efficiency = np.array([25, 30, 35, 40, 45])
average_efficiency = np.mean(fuel_efficiency)
percentage_improvement = ((fuel_efficiency[1] - fuel_efficiency[0]) /
                          fuel_efficiency[0]) * 100
print("Average fuel efficiency:", average_efficiency, "MPG")
print("Percentage improvement:", percentage_improvement, "%")
```
